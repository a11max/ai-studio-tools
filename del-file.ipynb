```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

folder_id = os.getenv('folder_id')
api_key = os.environ.get('api_key')
```

```python
client = OpenAI(
    api_key=api_key,
    base_url="https://ai.api.cloud.yandex.net/v1",
    project=folder_id,
    )
```

```python
list_resp = client.files.list()
files = list(list_resp.data)  # list_resp.data — это список объектов File
files
```

```python
for f in files:
    file_id = f.id
    try:
        deleted = client.files.delete(file_id)
        # В OpenAI API при успешном удалении возвращается объект с deleted=True
        if deleted.deleted:
            print(f"Файл {file_id} удалён")
        else:
            print(f"Файл {file_id} не помечен как удалённый (странный ответ)")
    except Exception as e:
        print(f"Ошибка при удалении файла {file_id}: {e}")
```

```python
list_vs_resp = client.vector_stores.list()
vector_stores = list(list_vs_resp.data)

print(f"Найдено поисковых индексов: {len(vector_stores)}")
for vs in vector_stores:
    print(f"ID: {vs.id} | Имя: {vs.name} | Файлов внутри: {vs.file_counts.total}")

vector_stores
```

```python
for vs in vector_stores:
    vs_id = vs.id
    try:
        deleted = client.vector_stores.delete(vs_id)
        if deleted.deleted:
            print(f"Поисковый индекс {vs_id} ({vs.name}) удалён")
        else:
            print(f"Индекс {vs_id} не помечен как удалённый: {deleted}")
    except Exception as e:
        print(f"Ошибка при удалении индекса {vs_id}: {e}")
```
