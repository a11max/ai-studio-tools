```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

folder_id = os.getenv('folder_id')
api_key = os.environ.get('api_key')
```

```python
client = OpenAI(
    api_key=api_key,
    base_url="https://ai.api.cloud.yandex.net/v1",
    project=folder_id,
    )
```

```python
list_resp = client.files.list()
files = list(list_resp.data)  # list_resp.data — это список объектов File
files
```

```python
for f in files:
    file_id = f.id
    try:
        deleted = client.files.delete(file_id)
        # В OpenAI API при успешном удалении возвращается объект с deleted=True
        if deleted.deleted:
            print(f"Файл {file_id} удалён")
        else:
            print(f"Файл {file_id} не помечен как удалённый (странный ответ)")
    except Exception as e:
        print(f"Ошибка при удалении файла {file_id}: {e}")
```
